## Establish applicability and decision roles

OCAP® is a registered trademark of FNIGC and a First Nations framework. This is a hypothetical First Nations-selected classroom exercise. Ownership concerns a collective relationship to information; control concerns decisions across its lifecycle; access includes First Nations access to their own information; possession concerns physical stewardship as a mechanism for asserting control. Consult FNIGC guidance and Nation-defined interpretations.


```python
applicability = {'teaching':True, 'scenario':'Hypothetical First Nations-selected application', 'actual_adoption_evidence':None, 'working_configuration_changed':False}
matrix = [
    dict(principle='Ownership', proposed_arrangement='Collective relationship recorded in agreement', decision_role='Fictional First Nation governing authority', evidence_needed='Nation-reviewed agreement', status='proposed'),
    dict(principle='Control', proposed_arrangement='Authority reviews collection, analysis, reuse and disposal', decision_role='Fictional designated governance body', evidence_needed='Lifecycle decision process and scoped decisions', status='proposed'),
    dict(principle='Access', proposed_arrangement='Nation staff can obtain usable data and review third-party requests', decision_role='Fictional data steward', evidence_needed='Tested access process and request log', status='proposed'),
    dict(principle='Possession', proposed_arrangement='Nation-controlled primary storage and keys; account for every copy', decision_role='Fictional infrastructure custodian', evidence_needed='Custody agreement and verified copy inventory', status='proposed'),
]
save_json('applicability.json', applicability)
save_csv('ocap_responsibility_matrix.csv', matrix)
```

## Inventory copies and identify possession gaps

Review primary data, backups, derived files, administrative credentials, key custody, export rights, and deletion verification. These are proposed arrangements only.


```python
copies = [
    {'copy_id':'primary','location':'Fictional Nation-controlled server','custodian':'Fictional Nation IT','key_holder':'Fictional Nation IT','export_available':True,'retention':'Nation-defined review','status':'proposed'},
    {'copy_id':'analyst-copy','location':'Fictional external workstation','custodian':'Fictional contractor','key_holder':'Fictional contractor','export_available':False,'retention':'undefined','status':'unresolved'},
    {'copy_id':'backup','location':'Fictional backup service','custodian':'Fictional service provider','key_holder':'unknown','export_available':False,'retention':'undefined','status':'unresolved'},
]
gaps = [dict(copy_id=c['copy_id'], gap='Resolve export, key custody, retention and verifiable return/deletion', owner='Fictional data steward') for c in copies if not c['export_available'] or c['retention']=='undefined' or c['key_holder']=='unknown']
assert len(gaps)==2
save_csv('custody_inventory.csv', copies)
save_csv('custody_remediation.csv', gaps)
print(gaps)
```

## Evaluate requests and simulate revocation

The local decision function models scope checks, including community access and reuse review. It does not enforce access or replace an authorized decision. Compare a Nation staff request with a contractor requesting new commercial use. Revocation must account for derivatives and backups; record uncertainty about removal rather than claiming it occurred.


```python
def review_request(request):
    gaps=[]
    if request['purpose']!='seasonal planning': gaps.append('New purpose requires review')
    if request['recipient']!='Nation planning staff': gaps.append('External recipient requires review')
    if not request.get('decision_evidence'): gaps.append('Missing scoped authorization evidence')
    return dict(request=request, status='hold' if gaps else 'ready for human evidence verification', gaps=gaps)
requests=[{'recipient':'Nation planning staff','purpose':'seasonal planning','decision_evidence':None}, {'recipient':'contractor','purpose':'commercial prediction','decision_evidence':None}]
decisions=[review_request(r) for r in requests]
assert decisions[0]['gaps']==['Missing scoped authorization evidence']
assert len(decisions[1]['gaps'])==3
save_json('access_request_decisions.json', decisions)
revocation = {'teaching':True, 'trigger':'Classroom simulation: contractor access revoked', 'immediate_actions':['Request account suspension','Stop transfers and downstream use','Notify fictional custodian through agreed process'], 'copies_to_review':[c['copy_id'] for c in copies], 'derivatives_to_review':['seasonal summary','cached export'], 'verification_needed':['Access logs','Custodian confirmation','Backup retention and deletion constraints'], 'status':'planned; no actual access changed', 'residual_risk':'Offline copies and backup deletion unverified'}
save_json('revocation_plan.json', revocation)
print(revocation)
```
