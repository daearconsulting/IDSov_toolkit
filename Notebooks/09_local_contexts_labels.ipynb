## Distinguish Labels, Notices, and access controls

Local Contexts TK and BC Labels support community expression of relationships and protocols. Indigenous communities customize Labels through the Hub; researchers and institutions use Notices to identify rights and interests. Labels do not replace consent, licenses, or technical access controls. Discuss TK Attribution, TK Seasonal, and BC provenance use cases from the primary references. A researcher must not invent community wording or select a Label on a community’s behalf.

## Prepare an integration boundary

This manifest is an internal teaching structure, not a Local Contexts API schema. In a real integration, retrieve the community-issued project information through the official Hub workflow and preserve titles, customized descriptions, and links according to usage guidance. Review what may be displayed before publication. No icons or official text are reproduced here.


```python
manifest = {'teaching':True, 'asset_id':'urn:example:idsov:narrative:v1', 'hub_project_url':None, 'community_issued_label':None, 'consultation_status':'not conducted', 'display_review':'pending', 'publication_status':'pending'}
def integration_gaps(m):
    gaps = []
    if not m.get('hub_project_url'): gaps.append('Missing verified Hub project link')
    label = m.get('community_issued_label')
    if not label: gaps.append('No community-issued Label supplied')
    else:
        for field in ['title','description','label_url','community_authority']:
            if not label.get(field): gaps.append('Missing Label '+field)
    if m.get('display_review')!='approved': gaps.append('Display review unresolved')
    return gaps
manifest['integration_gaps'] = integration_gaps(manifest)
assert len(manifest['integration_gaps']) == 3
save_json('local_contexts_integration_manifest.json', manifest)
print('\n'.join(manifest['integration_gaps']))
```

## Handle updates without silently losing protocols

A repository copy can become stale. Record source checks, changed project information, affected derivatives, and who reviews cached displays. Do not strip Label metadata when exporting or assume a Label grants publication permission.
