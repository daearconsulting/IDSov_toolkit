## Decision exercise

A researcher proposes a drought chart before a community question has been
established. Which decision belongs to the researcher and which belongs to the
Nation? Record the question as proposed and authorization as pending.


```python
import json
brief = {
    "nation": ctx.config.name,
    "question": "Proposed: how do drought-month summaries change when incomplete years are excluded?",
    "benefit": "Proposed: support discussion of environmental monitoring needs",
    "decision_authority": "Not yet documented",
    "technical_contributor": "Classroom analyst",
    "research_authorization": "pending",
    "sharing_authorization": "pending",
    "teaching": ctx.config.teaching,
}
(out/"00_project_brief.json").write_text(json.dumps(brief, indent=2), encoding="utf-8")
print(json.dumps(brief, indent=2))
```
