```python
import sys
from pathlib import Path

REPO_ROOT = Path.cwd().resolve()
while not (REPO_ROOT/"toolkit"/"__init__.py").exists():
    if REPO_ROOT.parent == REPO_ROOT:
        raise FileNotFoundError("Run from inside the IDSov_toolkit repository")
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from toolkit.teaching import load_context, output_directory
ctx = load_context(REPO_ROOT)
out = output_directory(REPO_ROOT, ctx)
print(f"Configuration: {ctx.config.name}")
print("Fictional teaching scenario" if ctx.config.teaching else "Working configuration; approvals require evidence")
```

## How the assessment works

Only selected frameworks are assessed. Unanswered questions withhold the overall
score. A complete score averages applicable framework scores equally, with
yes=1, partial=0.5, no=0; justified `na` answers are excluded. This weighting is a
teaching convention, not a validated measure of sovereignty or permission.

Every answered question needs an evidence reference, reviewer, and date to be
documented for review. Research and sharing approvals are separate required
gates even when a framework is disabled. Missing, denied, or revoked approvals
cannot be offset by a score. Teaching assessments never establish authorization.


```python
from toolkit.audit import SovereigntyAudit
audit = SovereigntyAudit(f"Draft assessment: {ctx.config.name}",
    reviewer="Classroom analyst", frameworks=ctx.config.active_frameworks, teaching=True)
print(audit.questions_list())
first = audit.questions[0]
audit.respond(first.id, "partial", notes="Teaching scenario: draft evidence has not been reviewed")
assert audit.overall_score is None
print(audit.report())
audit.save(out / "03_assessment.txt")
```

```python
import json
plan = [{"question_id": q.id, "action": "Gather evidence and obtain review",
         "owner": "To assign", "due_date": "", "evidence_needed": q.guidance,
         "status": "pending"} for q in audit.questions
        if q.response in (None, "no", "partial")]
(out / "03_remediation_plan.json").write_text(json.dumps(plan, indent=2), encoding="utf-8")
print(f"Remediation actions: {len(plan)}")
```
