## Frameworks inform Nation-defined governance

| Framework | Teaching focus | Source |
|---|---|---|
| OCAP® | Ownership, Control, Access, Possession; specifically First Nations principles | [FNIGC](https://fnigc.ca/ocap-training/) |
| CARE | Collective Benefit, Authority to Control, Responsibility, Ethics | [GIDA](https://www.gida-global.org/care) |
| FAIR | Findable, Accessible, Interoperable, Reusable | [FAIR principles](https://www.go-fair.org/fair-principles/) |
| IEEE 2890-2025 | Recommended Practice for Provenance of Indigenous Peoples' Data | [IEEE](https://standards.ieee.org/ieee/2890/10318/) |

OCAP® is a registered trademark of the First Nations Information Governance
Centre (FNIGC). Do not presume adoption by another Nation. OCAP is disabled in
the examples unless deliberately selected. Nation policies take precedence in
determining applicable requirements.

FAIR A1.2 allows authentication and authorization: restricted data can be FAIR.
Discoverability must also respect approved metadata disclosure rules.

The toolkit uses its own JSON schema. IEEE conformance has not been established.
See `docs/frameworks_reference.md` for a conceptual crosswalk and its limits.
Local Contexts needs a separate Nation-led review; the automated audit does not
provide a Local Contexts question set.
