```python
import sys
from pathlib import Path

REPO_ROOT = Path.cwd().resolve()
while not (REPO_ROOT/"toolkit"/"__init__.py").exists():
    if REPO_ROOT.parent == REPO_ROOT:
        raise FileNotFoundError("Run from inside the IDSov_toolkit repository")
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from toolkit.teaching import load_context, output_directory
ctx = load_context(REPO_ROOT)
out = output_directory(REPO_ROOT, ctx)
print(f"Configuration: {ctx.config.name}")
print("Fictional teaching scenario" if ctx.config.teaching else "Working configuration; approvals require evidence")
```

```python
import json
handoff = {
    "nation": ctx.config.name,
    "configuration_status": "teaching" if ctx.config.teaching else "awaiting review",
    "deliverables_present": sorted(p.name for p in out.iterdir() if p.is_file()),
    "steward": ctx.config.department,
    "reviewer": "", "review_date": "", "review_evidence": "",
    "research_authorization": "pending", "sharing_authorization": "pending",
    "retention_and_revocation": "To agree with designated steward",
    "return_of_findings": "Planned; receipt not yet documented",
}
(out / "06_handoff_manifest.json").write_text(json.dumps(handoff, indent=2), encoding="utf-8")
print(json.dumps(handoff, indent=2))
```
