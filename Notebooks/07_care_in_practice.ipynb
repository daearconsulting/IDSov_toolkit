## Translate principles into accountable actions

CARE connects Collective Benefit, Authority to Control, Responsibility, and Ethics to project decisions. These local teaching actions are not an exhaustive assessment of GIDA’s subprinciples. Community priorities determine the benefit; a researcher’s publication is not sufficient evidence of it. Separate a proposed action from evidence that it happened.


```python
actions = [
    dict(principle='Collective Benefit', decision='Prioritize seasonal water planning', owner='Fictional community planning team', evidence_needed='Community-reviewed question and benefit plan', status='proposed', review_trigger='Before collection'),
    dict(principle='Authority to Control', decision='Require steward review for each reuse', owner='Fictional designated authority', evidence_needed='Scoped decision with review and withdrawal terms', status='pending', review_trigger='New purpose or recipient'),
    dict(principle='Responsibility', decision='Budget paid training and return usable tables', owner='Fictional project lead', evidence_needed='Training budget and delivery receipt', status='proposed', review_trigger='Each reporting cycle'),
    dict(principle='Ethics', decision='Assess risks of identifying small sampling areas', owner='Fictional review group', evidence_needed='Risk review and community-defined mitigation', status='pending', review_trigger='Before any disclosure'),
]
save_csv('care_action_register.csv', actions)
for row in actions: print(row['principle'], '->', row['decision'], '('+row['status']+')')
```

## Evaluate scope before reuse

Discuss: an analyst receives a public upstream dataset. Can the combined community interpretation be sold to a vendor? Evaluate purpose, recipients, benefits, and decision evidence separately. The function below is a classroom triage rule, not a CARE compliance algorithm or permission service.


```python
agreement = {'purposes':['seasonal planning'], 'recipients':['community planning team'], 'decision_evidence':None}
def triage(request, agreement):
    gaps = []
    if request['purpose'] not in agreement['purposes']: gaps.append('Purpose outside proposed scope')
    if request['recipient'] not in agreement['recipients']: gaps.append('Recipient outside proposed scope')
    if not agreement['decision_evidence']: gaps.append('Decision evidence missing')
    return {'request':request, 'status':'hold for review' if gaps else 'ready for human evidence review', 'gaps':gaps}
requests = [dict(purpose='seasonal planning', recipient='community planning team'), dict(purpose='commercial risk scoring', recipient='external vendor')]
decisions = [triage(r, agreement) for r in requests]
assert len(decisions[1]['gaps']) == 3
assert all(d['status']=='hold for review' for d in decisions)
save_json('reuse_decisions.json', decisions)
print(json.dumps(decisions, indent=2))
```

## Plan benefit delivery and respond to changed conditions

Choose who can confirm usefulness and what happens when the community changes priorities. An emailed report does not establish that findings were returned in an accessible, useful form.


```python
return_plan = {'teaching':True, 'product':'Seasonal summary plus plain-language interpretation', 'format':['CSV','large-print briefing'], 'recipient':'Fictional planning team', 'delivery_status':'not delivered', 'success_measure':'Planning team confirms usefulness for its chosen question', 'feedback_owner':'Fictional community liaison', 'review_trigger':'Changed priorities or concerns about harm', 'response':'Pause reuse, consult decision authority, revise or withdraw draft', 'budget_status':'unfunded proposal'}
save_json('return_of_findings_plan.json', return_plan)
print(return_plan)
```
